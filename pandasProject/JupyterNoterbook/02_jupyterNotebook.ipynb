- apply()函数 --> 他可以自定义函数

```python
# 背景:目前我们用的都是Pandas提供好的函数，例如:count()max(),min()...
# 如果要传入一些自定义的逻辑，此时就需要用到 appLy()函数了
# 格式:df.apply(自定义函数，axis=0)  #0一列，1一行
```

```python
# 需求:同时 获取到 多列的最大值 和 最小值的差值.例如:open列cLose列
df.open.max() - df.close.min()

# 思路1：分解版
def my_func(col):
    return col.max() - col.min()
# 2.获取open和close列
df[['open','close']]   #找到指定列
df.iloc[:,[0,1]]
df.loc[:,['open','close']]

# 3.通过applv()函数.调用上述的白定义函数.作用到 指定的列
df[['open','close']].apply(my_func,axis=1) # axis=1,对列进行操作
df[['open','close']].apply(my_func,axis=0)
df[['open','close']].apply(my_func) # 效果同上、 axis=0，默认值，底层 -->把df的各列分别传入到my_func函数中，并返回结果

```
