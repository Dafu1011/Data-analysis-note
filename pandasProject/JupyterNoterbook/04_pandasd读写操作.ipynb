# 1.Pandas操作csv文件

```python
# 1. 读取csv文件，获取数据
# 参数解释：参数1：文件路径，参数2：分隔符，参数3：指定列
df = pd.read_csv('./../Data/stock_data.csv', sep=',',usecols=['open','high','low','close','volume']) # 默认使用,分隔符，获取数据指定列
df
```

```python
# 2.把读取的数据，写到文件中
df[:10].to_csv('./../Data/my_file1.csv',sep=',',index=False) # index=False表示不写入索引列
print("写入成功！")
```

```python
# 3.特殊的csv文件 ---> tsv文件
# 区别：csv文件 --> 以 ， 做分隔    tsv文件 --> 以  \t （tab键）做分隔
df[:5].to_csv('./../Data/my_file2.csv',sep='\t',index=False) # index=False表示不写入索引列
print("写入成功！")
```

```python
# 4.读取tsv文件
# 参数解释：参数1：文件路径，参数2：分隔符，参数3：指定索引列
df2 = pd.read_csv('./../Data/my_file2.csv',sep='\t',index_col=0)
df2
```

# 2.Panndas操作Mysql数据表

```python
# 前提：Anaconda已经装入 pymysql 和 sqlalchemy 模块
# 安装方式：pip install pymysql
# 安装方式：pip install sqlalchemy
# 此外还要装入 mysql数据库
```

```python
# 需求：把数据写入MySQL数据库表中
# 1.准备写到mysql数据表 的数据
data = pd.read_csv('./../Data/cvs文件.cvs',encoding='utf-8',index_col=0)
data
```

```python
# 2.导包
# from sqlalchemy import create_engine # 导入引擎对象

```

```python
# 3.创建引擎对象
# 参数解释：参数1：数据库类型+数据库驱动名://用户名:密码@数据库地址:端口号/数据库名?编码方式
engine = create_engine('mysql+pymysql://root:123456@localhost:3306/python?charset=utf8')

# # 4.具体向数据路中写入数据
# # 参数解释：参数1：数据表名，参数2：引擎对象，参数3：是否把索引列写入数据表，参数4：如果数据表存在 如何处理
data.to_sql('my_table1',engine,index=False,if_exists='append')
print("写入成功！")
```

```python
# 6.从sql表中读取数据、
sql_df =pd.read_sql('my_table1',engine)
sql_df =pd.read_sql('select name,AkA from my_table1 limit 0,2',engine)
sql_df
```
