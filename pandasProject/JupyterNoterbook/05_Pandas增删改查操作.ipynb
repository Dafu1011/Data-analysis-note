```python
# 2.从原始数据集中，拷贝一部分出来，做操作
df_new = df[:5].copy()
df_new
```

```python
# 3.新增列
# 思路1：通过直接赋值的方式 新增1列
df_new['c1'] = 66   # 固定写法
df_new['c2'] = ['12','23','哈哈哈','ae','3']  # 传入列表
df_new['c3'] = df_new.year * 2  # 通过已有列（Series对象）计算

def fun1():
    return '88888'

df_new['c4'] = fun1()
df_new
```

```python
# 1.准备数据
df4 = df[:10].copy()
df4
```

```python
# 2.删除行 drop()函数 ，除非指定inplace=True，否则不会修改原数据
# df4.drop(index=[0,2,4])  # 不会修改原数据
df4.drop(index=[0], inplace=True) #会修改元数据

df4
```

```python
# 8. rank()，类似于SQL的窗口函数
df8 = pd.DataFrame({
    '姓名':['小明','小美','小强','小兰'],
    '成绩':[100,90,90,70]
})
df8
```

```python
df8['成绩排名_min'] = df8.成绩.rank(method='min',ascending=False)
df8['成绩排名_max'] = df8.成绩.rank(method='max',ascending=False)
df8['成绩排名_dense'] = df8.成绩.rank(method='dense',ascending=False)
df8['成绩排名_avg'] = df8.成绩.rank(method='average',ascending=False)
df8['成绩排名_默认'] = df8.成绩.rank(ascending=False)
df8
```
