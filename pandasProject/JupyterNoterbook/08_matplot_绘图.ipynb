## 2.1 准备数据并绘制折线图,

```python
# 1.准备x轴数据（60分钟） 和 y轴数据（6每分钟的温度） 的数据
x = range(60)  # x的值：0 ~ 60 包左不包右
y_beijing = [random.uniform(15,18) for i in x]
# print(y_beijing)

# 2.创建画布
plt.figure(figsize=(10,5),dpi=80)

# 3.绘图
plt.plot(x,y_beijing,color='g')

# 4.具体绘图动作
plt.show()
```

## 2.2 添加x轴，y轴的刻度信息

```python
# 涉及到的函数：plt.xticks(x,**kwargs) 和 plt.yticks(y,**kwargs)

# 1.准备x轴数据（60分钟） 和 y轴数据（6每分钟的温度） 的数据
x = range(60)  # x的值：0 ~ 60 包左不包右
y_beijing = [random.uniform(15,18) for i in x]
# print(y_beijing)

# 2.创建画布
plt.figure(figsize=(10,5),dpi=80)

# 3.绘图
plt.plot(x,y_beijing,color='g')

# 4.添加x轴，y轴的刻度信息
# 4.1 准备x轴刻度信息（x轴刻度值对应的标签）
x_ticks_label = [f'11点{i}分' for i in x]
y_ticks = range(40)

# 4.2具体的添加x轴和y轴信息的动作     [::5]  步长为5
plt.xticks(x[::5],x_ticks_label[::5]) # 参1：x轴刻度 参2：x轴刻度值对应的标签
plt.yticks(y_ticks[::5]) # 参1：y轴刻度 参2：y轴刻度信息

# 5.添加网格
# 参1：网格线的样式 -  --  -.  :
# 参2：透明度 alpha
plt.grid(linestyle='-.', alpha=0.5)

# 6.具体绘图动作
plt.show()
# plt.show()放到最后
# plt.showO会释放figure资源如果在显示图像之后保存图片将只能保存空图片。
```

## 2.3 添加标题 x轴，y轴的标签

```python
# 涉及到的函数：plt.xticks(x,**kwargs) 和 plt.yticks(y,**kwargs)

# 1.准备x轴数据（60分钟） 和 y轴数据（6每分钟的温度） 的数据
x = range(60)  # x的值：0 ~ 60 包左不包右
y_beijing = [random.uniform(15,18) for i in x]
# print(y_beijing)

# 2.创建画布
plt.figure(figsize=(10,5),dpi=80)

# 3.绘图
plt.plot(x,y_beijing,color='g')

# 4.添加x轴，y轴的刻度信息
# 4.1 准备x轴刻度信息（x轴刻度值对应的标签）
x_ticks_label = [f'11点{i}分' for i in x]
y_ticks = range(40)

# 4.2具体的添加x轴和y轴信息的动作     [::5]  步长为5
plt.xticks(x[::5],x_ticks_label[::5]) # 参1：x轴刻度 参2：x轴刻度值对应的标签
plt.yticks(y_ticks[::5]) # 参1：y轴刻度 参2：y轴刻度信息

# 5.添加网格
# 参1：网格线的样式 -  --  -.  :
# 参2：透明度 alpha
plt.grid(linestyle='-.', alpha=0.5)

# 6.添加标题
plt.title('北京某天1小时内温度折线图',fontsize=20)

# 7.x轴，y轴的标签
plt.xlabel('时间',fontsize=12)
plt.ylabel('温度',fontsize=12)

# 8.保存图片
plt.savefig('./Img/beijing_1hour_temp.png',dpi=80)

# 9.具体绘图动作
plt.show()
# plt.show()放到最后
# plt.showO会释放figure资源如果在显示图像之后保存图片将只能保存空图片。
```

# 4.一个画布，多和坐标轴 --> 面向对象画图

```python
# 1.准备数据
x = range(60)
y_bejing = [random.uniform(5,18) for i in x]
y_shanghai = [random.uniform(10,15) for i in x]

# 2.创建画布
# plt.figure(figsize=(20,8),dpi=80)
# 以元组的形式返回结果
fig,axes = plt.subplots(nrows=1,ncols=2,figsize=(20,8),dpi=80)

# 3.具体的绘制动作
# 3.1 绘制出两个折线
# plt.plot(x,y_bejing,color='red',label='北京')
# plt.plot(x,y_shanghai,color='green',label='上海',linestyle='--')
axes[0].plot(x,y_bejing,color='pink',label='北京')
axes[1].plot(x,y_shanghai,color='green',label='上海',linestyle='--')

# 3.2 创建x轴的刻度值 和 Y轴的刻度值范围
x_ticks_label = [f'11点{i}分' for i in x]
y_ticks = range(20)
# plt.xticks(x[::5],x_ticks_label[::5])
# plt.yticks(y_ticks[::2])
# 设置 左侧坐标轴
axes[0].set_xticks(x[::5],x_ticks_label[::5])
axes[0].set_yticks(y_ticks[::2])
# 设置 右侧坐标轴
axes[1].set_xticks(x[::5],x_ticks_label[::5])
axes[1].set_yticks(y_ticks[::2])


# 3.3 标题，及 x轴 ，y轴标签
# plt.title('北京上海1小时内的气温折线图',fontsize=20)
# plt.xlabel('时间',fontsize=12)

axes[0].set_title('北京1小时内的气温折线图',fontsize=20)
axes[0].set_xlabel('时间',fontsize=12)

axes[1].set_title('上海1小时内的气温折线图',fontsize=20)
axes[1].set_xlabel('时间',fontsize=12)

# 3.4 添加网格
axes[0].grid(linestyle='--',alpha=0.3)
axes[1].grid(linestyle='--',alpha=0.3)

# 3.5 添加图例
# plt.legend(loc='best')  # upper left 、 upper right ...
# plt.legend(loc=0) # 效果同上  1、2、 3、4、...(0~10)

axes[0].legend(loc='best')  # upper left 、 upper right ...
axes[0].legend(loc=0) # 效果同上  1、2、 3、4、...(0~10)

axes[1].legend(loc='best')  # upper left 、 upper right ...
axes[1].legend(loc=0) # 效果同上  1、2、 3、4、...(0~10)

# 4.保存图片
plt.savefig('./Img/picture.png')

# 5.绘制图形
plt.show()


```

```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib import cbook, cm
from matplotlib.colors import LightSource

# Load and format data
dem = cbook.get_sample_data('jacksboro_fault_dem.npz')
z = dem['elevation']
nrows, ncols = z.shape
x = np.linspace(dem['xmin'], dem['xmax'], ncols)
y = np.linspace(dem['ymin'], dem['ymax'], nrows)
x, y = np.meshgrid(x, y)

region = np.s_[5:50, 5:50]
x, y, z = x[region], y[region], z[region]

# Set up plot
fig, ax = plt.subplots(subplot_kw=dict(projection='3d'))

ls = LightSource(270, 45)
# To use a custom hillshading mode, override the built-in shading and pass
# in the rgb colors of the shaded surface calculated from "shade".
rgb = ls.shade(z, cmap=cm.gist_earth, vert_exag=0.1, blend_mode='soft')
surf = ax.plot_surface(x, y, z, rstride=1, cstride=1, facecolors=rgb,
                       linewidth=0, antialiased=False, shade=False)

plt.show()
```

# 6. Matplotlib 绘制其他常用图形

## 6.1 柱状图 bar() --> 适用于 对比，统计...

```python
# 0. 准备数据
x = ['A','B','C','D']
y = [10,5,30,21]

# 1. 创建画布
plt.figure(figsize=(20,8),dpi=80)

# 2. 绘制柱状图
plt.bar(x,y)
# 2.1 设置网格
plt.grid(linestyle='--',alpha=0.3)
# 2.2 添加标题
plt.title('柱状图',fontsize=20)
# 2.3 添加x轴，y轴标签
plt.xlabel('类别',fontsize=12)
plt.ylabel('数量',fontsize=12)

# 3.显示图像
plt.show()
```

## 6.1 直方图 hist() --> 适用于 连续数据的 分布，统计...

```python
# 0. 准备数据
data = np.random.randn(500)  # 生成500个服从正太分布的随机数

# 1. 创建画布
plt.figure(figsize=(20,8),dpi=80)

# 2. 绘制直方图
plt.hist(data,bins=5) # bins=5 表示5个柱
# 2.1 设置网格
plt.grid(linestyle='--',alpha=0.3)
# 2.2 添加标题
plt.title('直方图',fontsize=20)
# 2.3 添加x轴，y轴标签
plt.xlabel('类别',fontsize=12)
plt.ylabel('数量',fontsize=12)

# 3.显示图像
plt.show()
```

## 6.3 饼图 pie() --> 适用于 比例，占比...

```python
sizes =[25,35,2,15]
labels = ['Category A', 'Category B', 'Category C', 'Category D']

# 1. 创建画布，绘制饼图
plt.pie(sizes, labels=labels, autopct='%1.1f%%') # 绘制饼图,显示百分比
plt.title("Ssimple Pie Chart")#设置图表标题
plt.show() #显示图表
```
