# 1.Pandas进阶语法_缺失值处理

## 1.1 思路1：删除缺失值

```python
# 1.读取数据
movie_df = pd.read_csv('./../Data/movie_data.csv')
movie_df
```

```python
# 2. 查看数据介绍
movie_df.columns  # 查看所有的列名
movie_df.info()   # 查看数据的 基本信息（列名、数据类型、非缺失值的数量等）
movie_df.describe() # 查看数据的 描述性统计信息
```

```python
# 3.
movie_df.dropna() # 删除所有有缺失值的行，不会修改原数据，加入参数inplace=True即可，默认axis=0 是行，axis=1 是列
movie_df.dropna(axis=1) # 删除所有有缺失值的列
movie_df
```

## 1.2 思路2：填充缺失值

```python
# 1.查看数据
movie_df
```

```python
# 2.判断某列是否有缺失值
pd.isnull(movie_df) # 判断df对象的 每列的 每个值 是否为空（缺失值）
pd.notnull(movie_df) # 判断df对象的 每列的 每个值 是否不为空（非缺失值）
```

```python
# 3. 判断某列是否是 包含缺失值的列
np.all(pd.notnull(movie_df))
```

```python
# 4.填充缺失值
# 写法1：填充固定值
movie_df.fillna(888888).info()
```

```python
# 写法2：填充 每列的平均值
movie_df['Votes'].fillna(movie_df['Votes'].mean(), inplace=True)
movie_df['Revenue (Millions)'].fillna(movie_df['Revenue (Millions)'].mean(), inplace=True)
movie_df.info()
```

```python
# 5.for循环到的方式，使用 每列的平均值 来填充各列的缺失值
# 5.1 获取每个列名
for column in movie_df.columns:
    # 5.2 判断该列是否包含缺失值
    # movie_df[column] : 根据列名，找打df中的某一列 --> Series对象
    # pd.notnull(某列) ： 判断该列的每个值是否为非缺失值，True —> 不为空，False —> 为空（缺失值）
    # np.all() ： 判断所有元素是否为True
    if np.all(pd.notnull(movie_df[column])) == False:
        # 5.3 说明有缺失值
        print(column) # 输出列名
        # 5.4 获取该列的平均值
        mean_value = movie_df[column].mean()
        print(mean_value)
        # 5.5 用该列的平均值填充该列的缺失值
        movie_df[column].fillna(mean_value, inplace=True)
```

```python
# 6.查看处理后的结果
movie_df.info()
```

# 2.数据合并

## 2.1 思路1：concat()，能行合并，还是列合并

```python
# 1.准备数据
df = pd.read_csv('./../Data/全球GDP数据.csv', encoding='utf8')
df

df1 = df[:10]
df1

df2 = df[10:20]
df2
```

```python
# 2.通过 concat() 合并数据
#细节：列合并，默认参考 列名
new_df = pd.concat([df1, df2])  # 默认是列合并 axis=0，行合并：axis=1
new_df = pd.concat([df1, df2], axis=0)  # axis=0 是列合并，效果同上

#细节：行合并，默认参考 行索引（值）
new_df = pd.concat([df1, df2], axis=1)  # axis=1 是行合并 有空值是列索引不一致所造成的
new_df
```

```python
# 3.修改df2的列索引
df2.index = [0,1,2,3,4,10,11,12,13,14] # 修改列索引,使得前5行列索引一致，做试验
# 4.再次合并
new_df = pd.concat([df1, df2], axis=1)
new_df
```

```python
new_df = pd.concat([df1, df2], axis=1,join='outer') #默认是：满外连接，即：左表全集 + 右表全集 + 交集

new_df = pd.concat([df1, df2], axis=1,join='inner') # 指定为：内连接，即：只要交集

new_df
```

## 2.2 思路2：merge() --> 只能进行 行合并（水平合并）

```python
# 准备数据集
df1 = pd.DataFrame({
    'key1': ['K0', 'K0', 'K1', 'K2'],
    'key2': ['K0', 'K1', 'K0', 'K1'],
    'A': ['A0', 'A1', 'A2', 'A3'],
    'B': ['B0', 'B1', 'B2', 'B3']
})

df2 = pd.DataFrame({
    'key1': ['K0', 'K1', 'K1', 'K2'],
    'key2': ['K0', 'K0', 'K0', 'K0'],
    'C': ['C0', 'C1', 'C2', 'C3'],
    'D': ['D0', 'D1', 'D2', 'D3']
})

# 2.查看数据
df1
df2
```

```python
# 3. 演示 merge()函数的 默认合并方式
pd.merge(df1, df2, how='inner',on=['key1', 'key2'])
pd.merge(df1, df2) # 效果同上，默认是 inner join 且参考同名列
```

```python
pd.merge(df1, df2, how='inner',on='key1')# 4.内连接，指定合并字段
```

```python
# 5. 外连接，指定合并字段
pd.merge(df1,df2,how='outer',on=['key1','key2']) #  满外连接，关联字段为：key1、key2
```

```python
# 6.左外连接，指定合并字段   左表全集 + 交集
pd.merge(df1,df2,how='left',on=['key1','key2'])

# 回顾：concat() 写法： pd.concat([df1,df2,df3,...]) 可以拼接多个，技能合并行，也能合并列，不能指定合并字段，默认是外连接合并方式
```

# 3. groupby()分组聚合 以及 分组过滤

```python
# 1. 准备数据
df = pd.read_csv('./../Data/uniqlo.csv')
df
```

## 3.1分组聚合

```python
# agg -> Aggregate(聚合的意思)
# 格式：df.groupby(['分组字段1','分组字段2',...]).agg({'列名1':'聚合函数名','列名2','聚合函数名'})
```

```python
# 1.场景1：按照单列分组
df.groupby(['city'])  # DataFrameGroupBy对象  -> DataFrame分组对象
df.groupby('city')    #细节：如果分组列只有一个，可以不加中括号
```

```python
# 2.场景2：按照多列分组
df.groupby(['city','channel'])
```

```python
# 3. 场景3：如何获取某个分组的数据
df.groupby(['city','channel']).get_group(('北京','门店购买'))

df.groupby(['city','channel']).get_group(('上海','网购'))
```

```python
# 4. 场景4：分组 + 聚合（聚合字段只有一个）
# 需求：根据 城市 和 渠道，对 销售金额 进行求和
# 写法1：通用版
df.groupby(['city','channel']).agg({'revenue':'sum'}) # DataFrame对象
# 写法2：简写版
df.groupby(['city','channel']).revenue.sum() #Series 对象
```

```python
# 5.场景5： 分组 + 聚合（聚合字段有两个，聚合字段相同）
# 需求：根据城市 和 销售渠道分组，计算：销售金额,订单数量  总和
# 写法1：通用版
df.groupby(['city','channel']).agg({'revenue':'sum','order':'sum'})

# 写法2：变形版
df.groupby(['city','channel'])[['revenue','order']].sum()
```

```python
# 6.场景6： 分组 + 聚合（聚合字段有两个，聚合字段不同）
# 需求：根据城市 和 销售渠道分组，分别计算：销售金额的平均值，成本的总和
df.groupby(['city','channel']).agg({'revenue':'mean','unit_cost':'sum'})

```

## 3.2 分组过滤.

```python
# 需求：按照城市分组，查询每组销售金额平均值 大于200的全部数据
# 即：算出每组的销售金额，找出金额大于200的的组，然后显示这些组所有数据

# 1.根据城市分组，计算每个城市的销售金额的 平均值
df.groupby(['city']).revenue.mean()

# 2. 根据城市分组，查看上海分组的数据
df.groupby(['city']).get_group('上海')

# 3. 完成需求
df.groupby(['city']).filter(lambda x:x.revenue.mean()>90000)
df.groupby(['city']).revenue.filter(lambda x:x.mean()>90000)
```

# 4. 交叉表（了解） 和 透视表（掌握）

```python
# 1. 交叉表演示：
# 创建一个示例数据
data = {
    '性别': ['男', '女', '男', '女','女','男', '女', '男'],
    '购买': ['是', '否', '是', '是', '否', '是', '是', '否'],
}

df = pd.DataFrame(data)

# 2. 交叉表
crosstab = pd.crosstab(df['性别'], df['购买'])
print(crosstab)
```

```python
# 2.透视表
data = {
    '性别': ['男', '女', '男', '女','女','男', '女', '男'],
    '购买': ['是', '否', '是', '是', '否', '是', '是', '否'],
    '金额':[100,120,150,180,90,120,150,180]
}

df = pd.DataFrame(data)

df.pivot_table(index='性别', columns='购买', values='金额', aggfunc='mean')
```

```python
 # 3. 透视表案例：
 #3.1 准备数据
df = pd.read_csv('./../Data/uniqlo.csv')
df

# 3.2  需求：根据城市， 销售渠道分组，计算 销售金额  总和
df.groupby(['city','channel']).agg({'revenue':'sum'})
# df.groupby(['city','channel'],as_index=False).agg({'revenue':'sum'}) #不把分组字段当做行索引，而是当做列

# 写法2：pivot_table()透视表
df.pivot_table(index='city',columns='channel',values='revenue',aggfunc='sum')
```
