# 1.Numpy的属性介绍

## 1.1创建ndarray对象

```python
# numpy -->和数学操作相关的库，numpy的数组是多维的，并且是固定类型，numpy的数组是ndarray，主要用于：矩阵运算
# 1.导包
import numpy as np
from holoviews.plotting.bokeh.styles import font_size

#2.创建numpy的ndarray对象。
arr1 = np.arange(15).reshape(3,5)
print(arr1)
```

## 1.2 演示numpy的常用属性

```python
print(f"numpy的轴：{arr1.ndim}") #2
print(f"numpy的维度：{arr1.shape}") #（3,5）
print(f"numpy的元素类型{arr1.dtype}") #int64   64位 = 8字节
print(f"numpy的元素个数{arr1.size}") #15
print(f"numpy的元素占用字节数:{arr1.itemsize}") #8
print(f"numpy的元素类型：{type(arr1)}") #<class 'numpy.ndarray'>
```

# 2.创建Numpy的ndarrary对象

## 2.1 array()函数，把Python列表  -->ndarray对对象

```python
# 创建Python列表对象
my_list = [1,2,3,4,5]
print(type(my_list))  # <class 'list'>
arr2 = np.array(my_list)
print(type(arr2))# <class 'numpy.ndarray'>
print(arr2)
```

## 2.2 arange()函数,创建ndarray对象

```python
# arange(起始,结束,步长,dtype=类型)，类似于Python中的range
#1.创建ndarrary对象
arr3 = np.arange(0,10,2,dtype=np.int32)
print(arr3)
print(type(arr3))
print(f"元素类型：{arr3.dtype}")
```

## 2.3 随机数生成 ndarrary对象

```python
# 1. random.rand() 生成0.0 ~ 1.0之间的，包左不包右的，随机小数矩阵
arr4 = np.random.rand(3,5) # 3行5列
print(arr4)
```

```python
# 2. random.randint(起始值，结束值，size=(行,列)) 生成指定范围，包左不包右的，随机小数矩阵
arr5 = np.random.randint(0,10,size=(2,6)) # 0~10之间，2行6列
print(arr5)
print(type(arr5))
```

```python
#3.random.uniform(起始值，结束值，size=(行,列)) 创建指定范围的随机矩阵,包左不包右
arr6 = np.random.uniform(3,9,size=(2,6))
print(arr6)
print(type(arr6))
```

## 2.4 numpy的类型转换 astype()

```python
# 1.创还能ndarray对象
arr7 = np.arange(1,10,dtype=np.float64)
print(arr7)
print(f"元素类型：{arr7.dtype}")
```

```python
# 2.类型转换
arr8 = arr7.astype(np.int32)
print(arr8)
print(f"元素类型：{arr8.dtype}")
```

## 2.5创建等比数列

```python
#1.创建等比数列，参1：起始值(充当10的指数)，参2：结束值（充当10的指数)，参3：元素个数
arr9 = np.logspace(0,3,4)
# 2.打印ndarrary对象
print(arr9)
print(f"元素类型：{arr9.dtype}")
```

```python
#3.创建等比数列，参1：起始值充当2的指数)，参2：结束值充当2的指数)，参3：元素个数
arr9 = np.logspace(0, 3, 4,base=2,dtype=np.int32)
# 4.打印ndarrary对象
print(arr9)
print(f"元素类型：{arr9.dtype}")
```

## 2.6 创建等差数列

```python
#1. 创建等差数列。参数1：起始值，参2：结束值，参3：元素个数，参4：是否包含结束值，参5：元素类型
arr10 = np.linspace(1,10,5,endpoint=False,dtype=np.int32)
print(arr10)
print(f"元素类型：{arr10.dtype}")
```

# 4.NUmpy的矩阵运算

## 4.1 减法运算

```python
# 1.定义数组
arr1 = np.array([10,20,30,40])
arr2 = np.arange(4)
# 2.打印数组
print(f"arr1:{arr1}")
print(f"arr2:{arr2}")
```

```python
# 3.减法运算
print(arr1-arr2)
```

```python
# 4.加法运算
print(arr1+arr2)
```

## 4.2矩阵乘法

```python
# 1. 定义数组
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 =  np.array([[1,2,3],[4,5,6]])
# 2.打印数组
print(f"arr1:{arr1}")
print(f"arr2:{arr2}")
```

```python
# 3.矩阵乘法，行列数一致
print(arr1*arr2)
print(np.multiply(arr1,arr2))
```

```python
# 4.矩阵乘法，行列数不一致
# 1. 定义数组
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 =  np.array([[1,2],[5,6],[7,8]])
# 2.打印数组
print(f"arr1:{arr1}")
print(f"arr2:{arr2}")
```

```python
# 6.矩阵乘法 --> 行列数不一致
# print(arr1*arr2) # 报错，行列数不一致

print(arr1 @ arr2) # 语法糖， A:2行3列，B:3行2列 ，A列=B行，即可操作，结果是A行B列
print(np.dot(arr1,arr2))
print(arr1.dot(arr2)) # 效果同上
```

```python
import pandas as pd
import numpy as np
import os
import sys
import seaborn as sns
import matplotlib.pyplot as plt
# print(sys.executable)
os.chdir(r"D:\Git\Git\pandasProject") # 切换工作目录
```

```python
#1.Pandas绘图可以基于 df对象直接绘图，它的底层依赖 Matplotlib
# 1.加载数据，并设置第1列为索引列
df = pd.read_csv("winemag-data_first100k.csv", index_col=0)
df.head()
```

```python
# 2.查看数据集
df.info()
df.describe()
df.columns
```

```python
# 3.需求1：绘制图形。展示产葡萄酒最多的3个产地信息。
# 思路1：根据产地（province）分组，基于points列（也可以是其他不包含缺失值的列）进行count统计即可
df.groupby('province',as_index=False).points.count().sort_values(by='points',ascending=False).head(3)
# 思路2：value_counts() 值的个数，且会自动降序排列，它相当于groupby()+count()+sort_values()
# df.province.value_counts().head(3)
df.province.value_counts()
df['province'].value_counts()
df['province'].value_counts().head(3)
```

```python
# 4.绘制图形。展示产地最高的10个产地
df.province.value_counts().head(10).plot() # 折线图,语法糖

# df.province.value_counts().head(10).plot.line() # 折线图
df.province.value_counts().head(10).plot.bar(figsize=(10,5),color=['r','b','g','y'],fontsize=13)   # 柱状图  figsize=(宽,高), color=['颜色1','颜色2',...],颜色不够是循环的
```

```python
# 5.最终写法
# step1：定义变量，记录参数，字典形式
text_kwargs=dict(figsize=(10,5),color=['r','b','g','y'],fontsize=13)
# print(type(text_kwargs)) # <class 'dict'>
# step2：调用绘图方法
df.province.value_counts().head(10).plot.bar(**text_kwargs)
```
